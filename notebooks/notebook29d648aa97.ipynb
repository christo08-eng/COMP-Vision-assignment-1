```python
# Task 2: Image Deblurring
# Install required packages
!pip install scikit-image opencv-python-headless matplotlib numpy -q

import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import img_as_float, img_as_ubyte
from skimage.restoration import wiener, richardson_lucy
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim

print("All libraries imported successfully!")

# Dataset paths
BLUR_PATH = "/kaggle/input/datasets/rahulbhalley/gopro-deblur/gopro_deblur/blur"
SHARP_PATH = "/kaggle/input/datasets/rahulbhalley/gopro-deblur/gopro_deblur/sharp"

# Check dataset is accessible
print(f"Blur folders: {os.listdir(BLUR_PATH)}")
print(f"Sharp folders: {os.listdir(SHARP_PATH)}")
```

```python
# Full paths to images
BLUR_IMGS = "/kaggle/input/datasets/rahulbhalley/gopro-deblur/gopro_deblur/blur/images"
SHARP_IMGS = "/kaggle/input/datasets/rahulbhalley/gopro-deblur/gopro_deblur/sharp/images"

# Get list of images
blur_files = sorted(os.listdir(BLUR_IMGS))
sharp_files = sorted(os.listdir(SHARP_IMGS))

print(f"Total blur images: {len(blur_files)}")
print(f"Total sharp images: {len(sharp_files)}")
print(f"Sample filenames: {blur_files[:5]}")

# Load a sample pair
sample_blur = cv2.imread(os.path.join(BLUR_IMGS, blur_files[0]))
sample_sharp = cv2.imread(os.path.join(SHARP_IMGS, sharp_files[0]))

# Convert BGR to RGB for display
sample_blur = cv2.cvtColor(sample_blur, cv2.COLOR_BGR2RGB)
sample_sharp = cv2.cvtColor(sample_sharp, cv2.COLOR_BGR2RGB)

# Display side by side
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
axes[0].imshow(sample_blur)
axes[0].set_title("Blurred Image", fontsize=14)
axes[0].axis("off")

axes[1].imshow(sample_sharp)
axes[1].set_title("Sharp (Ground Truth)", fontsize=14)
axes[1].axis("off")

plt.suptitle("Sample Image Pair from GoPro Dataset", fontsize=16)
plt.tight_layout()
plt.show()

print(f"Image shape: {sample_blur.shape}")
```

```python
from skimage.restoration import wiener
from skimage import img_as_float

def apply_wiener(blur_img, psf_size=15, balance=0.05):
    """Apply Wiener filter to each channel separately"""
    blur_float = img_as_float(blur_img)
    result = np.zeros_like(blur_float)
    
    # Larger PSF to better model motion blur
    psf = np.ones((psf_size, psf_size)) / (psf_size ** 2)
    
    for i in range(3):
        result[:,:,i] = wiener(blur_float[:,:,i], psf, balance=balance)
    
    result = np.clip(result, 0, 1)
    return result

# Try multiple PSF sizes to find best result
blur_float = img_as_float(sample_blur)
sharp_float = img_as_float(sample_sharp)

best_psnr = 0
best_result = None
best_size = 0

for psf_size in [5, 10, 15, 20, 25]:
    result = apply_wiener(sample_blur, psf_size=psf_size, balance=0.05)
    p = psnr(sharp_float, result)
    s = ssim(sharp_float, result, channel_axis=2, data_range=1.0)
    print(f"PSF size {psf_size:2d} — PSNR: {p:.2f} dB  SSIM: {s:.4f}")
    if p > best_psnr:
        best_psnr = p
        best_result = result
        best_size = psf_size

print(f"\nBest PSF size: {best_size} with PSNR: {best_psnr:.2f} dB")

# Display best result
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
axes[0].imshow(sample_blur)
axes[0].set_title(f"Blurred\nPSNR: {psnr(sharp_float, blur_float):.2f} dB", fontsize=12)
axes[0].axis("off")

axes[1].imshow(best_result)
axes[1].set_title(f"Wiener Deblurred (PSF={best_size})\nPSNR: {best_psnr:.2f} dB", fontsize=12)
axes[1].axis("off")

axes[2].imshow(sample_sharp)
axes[2].set_title("Sharp (Ground Truth)", fontsize=12)
axes[2].axis("off")

plt.suptitle("Wiener Filter — Best Result", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
from skimage.restoration import richardson_lucy
from skimage import img_as_float

def apply_rl(blur_img, psf_size=5, iterations=10):
    """Apply Richardson-Lucy deconvolution to each channel"""
    blur_float = img_as_float(blur_img)
    result = np.zeros_like(blur_float)
    
    psf = np.ones((psf_size, psf_size)) / (psf_size ** 2)
    
    for i in range(3):
        result[:,:,i] = richardson_lucy(blur_float[:,:,i], psf, num_iter=iterations)
    
    result = np.clip(result, 0, 1)
    return result

blur_float = img_as_float(sample_blur)
sharp_float = img_as_float(sample_sharp)

print("=== Richardson-Lucy Results ===")
best_psnr = 0
best_result = None
best_params = {}

for psf_size in [3, 5, 7]:
    for iters in [5, 10, 20, 30]:
        result = apply_rl(sample_blur, psf_size=psf_size, iterations=iters)
        p = psnr(sharp_float, result)
        s = ssim(sharp_float, result, channel_axis=2, data_range=1.0)
        print(f"PSF={psf_size} Iters={iters:2d} — PSNR: {p:.2f} dB  SSIM: {s:.4f}")
        if p > best_psnr:
            best_psnr = p
            best_result = result
            best_params = {'psf_size': psf_size, 'iters': iters}

print(f"\nBest params: {best_params}")
print(f"Best PSNR:   {best_psnr:.2f} dB")
print(f"Blurred PSNR: {psnr(sharp_float, blur_float):.2f} dB")

# Display
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
axes[0].imshow(sample_blur)
axes[0].set_title(f"Blurred\nPSNR: {psnr(sharp_float, blur_float):.2f} dB", fontsize=12)
axes[0].axis("off")

axes[1].imshow(best_result)
axes[1].set_title(f"RL Deblurred\nPSNR: {best_psnr:.2f} dB", fontsize=12)
axes[1].axis("off")

axes[2].imshow(sample_sharp)
axes[2].set_title("Sharp (Ground Truth)", fontsize=12)
axes[2].axis("off")

plt.suptitle("Richardson-Lucy Deconvolution Results", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
import torch
import torch.nn.functional as F
from skimage import img_as_float
import numpy as np

# Load model
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

model = MPRNet()
checkpoint = torch.load(
    '/kaggle/working/MPRNet/Motion_Deblurring/pretrained_models/model_deblurring.pth',
    map_location=device)
model.load_state_dict(checkpoint['state_dict'])
model.to(device)
model.eval()
print("Model loaded!")

def deblur_image(model, image, device):
    img = torch.from_numpy(img_as_float(image).transpose(2,0,1)).float()
    img = img.unsqueeze(0).to(device)
    h, w = img.shape[2], img.shape[3]
    pad_h = (8 - h % 8) % 8
    pad_w = (8 - w % 8) % 8
    img = F.pad(img, (0, pad_w, 0, pad_h), 'reflect')
    with torch.no_grad():
        output = model(img)[0]
    output = output[:, :, :h, :w]
    output = output.squeeze(0).cpu().numpy().transpose(1,2,0)
    output = np.clip(output, 0, 1)
    return output

# Run on sample image
print("Deblurring sample image, please wait...")
blur_float = img_as_float(sample_blur)
sharp_float = img_as_float(sample_sharp)

deblurred_dl = deblur_image(model, sample_blur, device)

# Metrics
p_blur = psnr(sharp_float, blur_float)
p_dl = psnr(sharp_float, deblurred_dl)
s_blur = ssim(sharp_float, blur_float, channel_axis=2, data_range=1.0)
s_dl = ssim(sharp_float, deblurred_dl, channel_axis=2, data_range=1.0)

print(f"\Results ")
print(f"Blurred    — PSNR: {p_blur:.2f} dB  SSIM: {s_blur:.4f}")
print(f"MPRNet DL  — PSNR: {p_dl:.2f} dB  SSIM: {s_dl:.4f}")

# Display
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
axes[0].imshow(sample_blur)
axes[0].set_title(f"Blurred\nPSNR: {p_blur:.2f} dB", fontsize=12)
axes[0].axis("off")
axes[1].imshow(deblurred_dl)
axes[1].set_title(f"MPRNet Deblurred\nPSNR: {p_dl:.2f} dB", fontsize=12)
axes[1].axis("off")
axes[2].imshow(sample_sharp)
axes[2].set_title("Sharp (Ground Truth)", fontsize=12)
axes[2].axis("off")

plt.suptitle("Deep Learning Deblurring — MPRNet", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
def run_detection(model, img_path, conf_threshold=0.25):
    """Run YOLOv8 detection and return results"""
    results = model(img_path, conf=conf_threshold, verbose=False)
    return results[0]

def draw_detections(image, results):
    """Draw bounding boxes on image"""
    img_draw = image.copy()
    boxes = results.boxes
    
    for box in boxes:
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        conf = float(box.conf[0])
        cls = int(box.cls[0])
        label = f"{results.names[cls]} {conf:.2f}"
        
        cv2.rectangle(img_draw, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(img_draw, label, (x1, y1-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
    return img_draw

# Load sample image pair
sample_idx = 0
blur_img = cv2.imread(os.path.join(BLUR_IMGS, blur_files[sample_idx]))
deblur_img = cv2.imread(os.path.join(DEBLURRED_PATH, deblurred_files[sample_idx]))
sharp_img = cv2.imread(os.path.join(SHARP_IMGS, sharp_files[sample_idx]))

blur_img = cv2.cvtColor(blur_img, cv2.COLOR_BGR2RGB)
deblur_img = cv2.cvtColor(deblur_img, cv2.COLOR_BGR2RGB)
sharp_img = cv2.cvtColor(sharp_img, cv2.COLOR_BGR2RGB)

# Save temp images for YOLO
cv2.imwrite('/kaggle/working/temp_blur.jpg', cv2.cvtColor(blur_img, cv2.COLOR_RGB2BGR))
cv2.imwrite('/kaggle/working/temp_deblur.jpg', cv2.cvtColor(deblur_img, cv2.COLOR_RGB2BGR))
cv2.imwrite('/kaggle/working/temp_sharp.jpg', cv2.cvtColor(sharp_img, cv2.COLOR_RGB2BGR))

# Run detection on all three
results_blur = run_detection(model_yolo, '/kaggle/working/temp_blur.jpg')
results_deblur = run_detection(model_yolo, '/kaggle/working/temp_deblur.jpg')
results_sharp = run_detection(model_yolo, '/kaggle/working/temp_sharp.jpg')

# Print detection summary
print("Detection Results — Sample Image")
print(f"Blurred    — Objects detected: {len(results_blur.boxes)}  "
      f"Avg confidence: {float(results_blur.boxes.conf.mean()):.3f}" 
      if len(results_blur.boxes) > 0 else "Blurred Objects detected: 0")
print(f"Deblurred  — Objects detected: {len(results_deblur.boxes)}  "
      f"Avg confidence: {float(results_deblur.boxes.conf.mean()):.3f}"
      if len(results_deblur.boxes) > 0 else "Deblurred Objects detected: 0")
print(f"Sharp      — Objects detected: {len(results_sharp.boxes)}  "
      f"Avg confidence: {float(results_sharp.boxes.conf.mean()):.3f}"
      if len(results_sharp.boxes) > 0 else "Sharp Objects detected: 0")

# Draw detections
blur_drawn = draw_detections(blur_img, results_blur)
deblur_drawn = draw_detections(deblur_img, results_deblur)
sharp_drawn = draw_detections(sharp_img, results_sharp)

# Display
fig, axes = plt.subplots(1, 3, figsize=(20, 7))
axes[0].imshow(blur_drawn)
axes[0].set_title(f"Blurred\n{len(results_blur.boxes)} detections", fontsize=12)
axes[0].axis("off")

axes[1].imshow(deblur_drawn)
axes[1].set_title(f"Deblurred\n{len(results_deblur.boxes)} detections", fontsize=12)
axes[1].axis("off")

axes[2].imshow(sharp_drawn)
axes[2].set_title(f"Sharp\n{len(results_sharp.boxes)} detections", fontsize=12)
axes[2].axis("off")

plt.suptitle("YOLOv8 Detection blurred vs deblurred vs Sharp", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
from collections import defaultdict

blur_classes = defaultdict(list)
deblur_classes = defaultdict(list)
sharp_classes = defaultdict(list)

print("Running per-class analysis on 50 images")

for i in tqdm(range(50)):
    blur_img = cv2.imread(os.path.join(BLUR_IMGS, blur_files[i]))
    deblur_img = cv2.imread(os.path.join(DEBLURRED_PATH, deblurred_files[i]))
    sharp_img = cv2.imread(os.path.join(SHARP_IMGS, sharp_files[i]))

    cv2.imwrite('/kaggle/working/temp_blur.jpg', blur_img)
    cv2.imwrite('/kaggle/working/temp_deblur.jpg', deblur_img)
    cv2.imwrite('/kaggle/working/temp_sharp.jpg', sharp_img)

    r_blur = run_detection(model_yolo, '/kaggle/working/temp_blur.jpg')
    r_deblur = run_detection(model_yolo, '/kaggle/working/temp_deblur.jpg')
    r_sharp = run_detection(model_yolo, '/kaggle/working/temp_sharp.jpg')

    # Collect per-class confidences
    for box in r_blur.boxes:
        cls_name = r_blur.names[int(box.cls[0])]
        blur_classes[cls_name].append(float(box.conf[0]))

    for box in r_deblur.boxes:
        cls_name = r_deblur.names[int(box.cls[0])]
        deblur_classes[cls_name].append(float(box.conf[0]))

    for box in r_sharp.boxes:
        cls_name = r_sharp.names[int(box.cls[0])]
        sharp_classes[cls_name].append(float(box.conf[0]))

# Find classes detected in all three
all_classes = set(blur_classes.keys()) | set(deblur_classes.keys())
print(f"\n Per-class Detection Summary")
print(f"{'Class':<15} {'Blur count':<12} {'Deblur count':<14} {'Sharp count'}")
print("-" * 55)

class_data = []
for cls in sorted(all_classes):
    b_count = len(blur_classes[cls])
    d_count = len(deblur_classes[cls])
    s_count = len(sharp_classes[cls])
    print(f"{cls:<15} {b_count:<12} {d_count:<14} {s_count}")
    class_data.append((cls, b_count, d_count, s_count))

# Plot top classes
top_classes = sorted(class_data, key=lambda x: x[3], reverse=True)[:6]
cls_names = [x[0] for x in top_classes]
b_counts = [x[1] for x in top_classes]
d_counts = [x[2] for x in top_classes]
s_counts = [x[3] for x in top_classes]

x = np.arange(len(cls_names))
width = 0.25

fig, ax = plt.subplots(figsize=(14, 6))
ax.bar(x - width, b_counts, width, label='Blurred', color='gray')
ax.bar(x, d_counts, width, label='Deblurred', color='blue')
ax.bar(x + width, s_counts, width, label='Sharp', color='green')

ax.set_title('Per-class detection counts — Blurred vs Deblurred vs Sharp')
ax.set_ylabel('Total detections across 50 images')
ax.set_xticks(x)
ax.set_xticklabels(cls_names, rotation=15)
ax.legend()
plt.tight_layout()
plt.show()
```

```python
import time

print("Measuring inference latency")
n_trials = 20
blur_times, deblur_times, sharp_times = [], [], []

for i in range(n_trials):
    blur_img = cv2.imread(os.path.join(BLUR_IMGS, blur_files[i]))
    deblur_img = cv2.imread(os.path.join(DEBLURRED_PATH, deblurred_files[i]))
    sharp_img = cv2.imread(os.path.join(SHARP_IMGS, sharp_files[i]))

    cv2.imwrite('/kaggle/working/temp_blur.jpg', blur_img)
    cv2.imwrite('/kaggle/working/temp_deblur.jpg', deblur_img)
    cv2.imwrite('/kaggle/working/temp_sharp.jpg', sharp_img)

    # Time each
    start = time.time()
    model_yolo('/kaggle/working/temp_blur.jpg', verbose=False)
    blur_times.append(time.time() - start)

    start = time.time()
    model_yolo('/kaggle/working/temp_deblur.jpg', verbose=False)
    deblur_times.append(time.time() - start)

    start = time.time()
    model_yolo('/kaggle/working/temp_sharp.jpg', verbose=False)
    sharp_times.append(time.time() - start)

print(f"\n=== Inference Latency (avg over {n_trials} images) ===")
print(f"Blurred images:    {np.mean(blur_times)*1000:.1f} ms per image")
print(f"Deblurred images:  {np.mean(deblur_times)*1000:.1f} ms per image")
print(f"Sharp images:      {np.mean(sharp_times)*1000:.1f} ms per image")

# Plot
fig, ax = plt.subplots(figsize=(8, 5))
methods = ['Blurred', 'Deblurred', 'Sharp']
times = [np.mean(blur_times)*1000, 
         np.mean(deblur_times)*1000, 
         np.mean(sharp_times)*1000]
ax.bar(methods, times, color=['gray', 'blue', 'green'])
ax.set_title('YOLOv8 Inference Latency per Image')
ax.set_ylabel('Time (ms)')
for i, v in enumerate(times):
    ax.text(i, v + 0.5, f'{v:.1f}ms', ha='center', fontsize=11)
plt.tight_layout()
plt.show()
```
